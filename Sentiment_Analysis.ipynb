```python
from transformers import pipeline
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
review = "The storyline was amazing, but the ending was disappointing."

result = classifier(review)

print(result)
```
